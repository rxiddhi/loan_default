# 03 EDA\nPurpose: Explore risk patterns and generate visual evidence for business decisions and dashboard design.

```python
from pathlib import Path\nimport pandas as pd\nimport seaborn as sns\nimport matplotlib.pyplot as plt\nimport plotly.express as px\n\nROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd()\nCLEAN_PATH = ROOT / 'data' / 'processed' / 'loan_clean.csv'\nFIG_DIR = ROOT / 'reports' / 'figures'\nSUMMARY_OUT = ROOT / 'reports' / 'phase_04_eda_summary.csv'\nFIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(CLEAN_PATH)\ndf.head()
```

```python
sns.set_theme(style='whitegrid')\nplt.figure(figsize=(6,4))\nsns.countplot(data=df, x='Default')\nplt.title('Default Distribution')\nplt.tight_layout()\nplt.savefig(FIG_DIR / 'eda_default_distribution.png', dpi=160)\nplt.show()
```

```python
plt.figure(figsize=(7,4))\nsns.boxplot(data=df, x='Default', y='CreditScore')\nplt.title('Credit Score vs Default')\nplt.tight_layout()\nplt.savefig(FIG_DIR / 'eda_creditscore_vs_default.png', dpi=160)\nplt.show()
```

```python
segment_summary = df.groupby('LoanPurpose', as_index=False).agg(\n    loans=('LoanID','count'),\n    default_rate=('Default','mean'),\n    avg_income=('Income','mean'),\n)\nsegment_summary['default_rate'] = (segment_summary['default_rate'] * 100).round(3)\nsegment_summary = segment_summary.sort_values('default_rate', ascending=False)\nsegment_summary.to_csv(SUMMARY_OUT, index=False)\nsegment_summary
```
